```python
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize


text = 'India is a great country.'

tokens = word_tokenize(text)

stop_words = set(stopwords.words('english'))

filtered_tokens = [word for word in tokens if word.lower() not in stop_words]

print(stop_words , '\n\n')

filtered_text = ' '.join(filtered_tokens)

print(filtered_text)
```

```python
strings = df.loc[:5, 'text'].str.lower()
```

```python
print(strings)
```

```python
from nltk import word_tokenize

words_in_each_string = []

for each_string in strings:
  words_in_each_string.append(word_tokenize(each_string))


words_in_each_string
```

```python
original_words_list = []
```

```python
original_words_list = []
p_stemmed_list = []
s_stemmed_list = []

for each_word_list in words_in_each_string:
  original_words_list.extend(each_word_list)
  p_stemmed_list.extend([p_stemmer.stem(word) for word in each_word_list])
  s_stemmed_list.extend([s_stemmer.stem(word) for word in each_word_list])

data = pd.DataFrame(original_words_list, columns=['original_word'])
data['p_stemmed'] = p_stemmed_list
data['s_stemmed'] = s_stemmed_list

data
```
